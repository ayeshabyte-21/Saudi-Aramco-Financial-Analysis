```python
income['Revenue Growth'] = income['Revenue'].pct_change() * 100

income['Operating Income Growth'] = income['Operating Income'].pct_change() * 100

income['Net Income Growth'] = income['Net Income'].pct_change() * 100

income[['Year', 'Revenue Growth',
        'Operating Income Growth',
        'Net Income Growth']].round(2)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(income['Year'], income['Revenue'],
         marker='o', label='Revenue')

plt.plot(income['Year'], income['Operating Income'],
         marker='o', label='Operating Income')

plt.plot(income['Year'], income['Net Income'],
         marker='o', label='Net Income')

plt.title('Saudi Aramco Financial Performance (2021–2025)')
plt.xlabel('Year')
plt.ylabel('SAR Million')

plt.xticks(income['Year'])
plt.legend()
plt.grid(True)

plt.show()
```

```python
income['Operating Margin'] = (
    income['Operating Income'] / income['Revenue']
) * 100

income['Net Profit Margin'] = (
    income['Net Income'] / income['Revenue']
) * 100

income[['Year', 'Operating Margin',
        'Net Profit Margin']].round(2)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(income['Year'], income['Operating Margin'],
         marker='o', label='Operating Margin')

plt.plot(income['Year'], income['Net Profit Margin'],
         marker='o', label='Net Profit Margin')

plt.title('Saudi Aramco Profitability Trends (2021–2025)')
plt.xlabel('Year')
plt.ylabel('Margin (%)')

plt.xticks(income['Year'])
plt.legend()
plt.grid(True)

plt.show()
```

```python
balance[['Year', 'Total Assets',
         'Total Liabilities', 'Total Equity']].describe()
```

```python
balance['Asset Growth %'] = (
    balance['Total Assets'].pct_change() * 100
)

balance['Liability Growth %'] = (
    balance['Total Liabilities'].pct_change() * 100
)

balance['Equity Growth %'] = (
    balance['Total Equity'].pct_change() * 100
)

balance[['Year', 'Asset Growth %',
         'Liability Growth %',
         'Equity Growth %']].round(2)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(balance['Year'], balance['Total Assets'],
         marker='o', label='Total Assets')

plt.plot(balance['Year'], balance['Total Liabilities'],
         marker='o', label='Total Liabilities')

plt.plot(balance['Year'], balance['Total Equity'],
         marker='o', label='Total Equity')

plt.title('Saudi Aramco Balance Sheet Trends (2021–2025)')
plt.xlabel('Year')
plt.ylabel('SAR Million')
plt.xticks(balance['Year'])
plt.legend()
plt.grid(True)
plt.show()
```

```python
balance['Current Ratio'] = (
    balance['Current Assets'] /
    balance['Current Liabilities']
)

balance['Working Capital'] = (
    balance['Current Assets'] -
    balance['Current Liabilities']
)

balance[['Year', 'Current Assets',
         'Current Liabilities',
         'Current Ratio',
         'Working Capital']].round(2)
```

```python
plt.figure(figsize=(9, 5))

plt.plot(balance['Year'], balance['Current Ratio'],
         marker='o', label='Current Ratio')

plt.axhline(y=1, linestyle='--',
            label='Current Ratio = 1')

plt.title('Saudi Aramco Liquidity Trend (2021–2025)')
plt.xlabel('Year')
plt.ylabel('Current Ratio')
plt.xticks(balance['Year'])
plt.legend()
plt.grid(True)
plt.show()
```

```python
balance['Liabilities to Equity'] = (
    balance['Total Liabilities'] /
    balance['Total Equity']
)

balance['Equity Ratio'] = (
    balance['Total Equity'] /
    balance['Total Assets'] * 100
)

balance[['Year', 'Liabilities to Equity',
         'Equity Ratio']].round(4)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(balance['Year'],
         balance['Liabilities to Equity'],
         marker='o', label='Liabilities-to-Equity')

plt.plot(balance['Year'],
         balance['Equity Ratio'] / 100,
         marker='o', label='Equity Ratio')

plt.title('Saudi Aramco Capital Structure (2021–2025)')
plt.xlabel('Year')
plt.ylabel('Ratio')
plt.xticks(balance['Year'])
plt.legend()
plt.grid(True)
plt.show()
```

```python
balance['Average Assets'] = (
    (balance['Total Assets'] +
     balance['Total Assets'].shift(1)) / 2
)

balance['ROA'] = (
    income['Net Income'] /
    balance['Average Assets'] * 100
)

balance[['Year', 'Average Assets', 'ROA']].round(2)
```

```python
balance.loc[0, 'Average Assets'] = (
    (1914261 + balance.loc[0, 'Total Assets']) / 2
)

balance.loc[0, 'ROA'] = (
    income.loc[0, 'Net Income'] /
    balance.loc[0, 'Average Assets'] * 100
)

balance[['Year', 'Average Assets', 'ROA']].round(2)
```

```python
balance['Average Equity'] = (
    (balance['Total Equity'] +
     balance['Total Equity'].shift(1)) / 2
)

balance['ROE'] = (
    income['Net Income'] /
    balance['Average Equity'] * 100
)

balance[['Year', 'Average Equity', 'ROE']].round(2)
```

```python
balance.loc[0, 'Average Equity'] = (
    (1101094 + balance.loc[0, 'Total Equity']) / 2
)

balance.loc[0, 'ROE'] = (
    income.loc[0, 'Net Income'] /
    balance.loc[0, 'Average Equity'] * 100
)

balance[['Year', 'Average Equity', 'ROE']].round(2)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(balance['Year'], balance['ROA'],
         marker='o', label='ROA')

plt.plot(balance['Year'], balance['ROE'],
         marker='o', label='ROE')

plt.title('Saudi Aramco: ROA vs ROE (2021–2025)')
plt.xlabel('Year')
plt.ylabel('Return (%)')
plt.xticks(balance['Year'])
plt.legend()
plt.grid(True)
plt.show()
```

```python
profitability_analysis = balance[
    ['Year', 'ROA', 'ROE']
].copy()

profitability_analysis['ROA'] = (
    profitability_analysis['ROA'].round(2)
)

profitability_analysis['ROE'] = (
    profitability_analysis['ROE'].round(2)
)

profitability_analysis
```

This supports financial stability, preserves the ability to invest in future operations, and improves dividend sustainability.

Why Liquidity and Leverage Matter
Liquidity measures a company's ability to meet its short-term obligations, while leverage indicates the extent to which it relies on liabilities to finance its assets. Examining both helps assess financial stability, the ability to manage financial risks, and the company's capacity to meet its obligations.
